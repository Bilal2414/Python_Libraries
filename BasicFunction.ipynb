```python
#Funtion with parameters
def average(a,b):
    avgValue = (a+b)/2
    print("The average of a and b is:", avgValue)
#function call with Arguments
average(7, 10)
average(10, 40)
average(20, 35)
```

```python
#Funtion with default parameters
def average(a=11,b=22):
    avgValue = (a+b)/2
    print("The average of a and b is:", avgValue)
#function call with Arguments
average(7, 10)
#Function call with default arguments
average()
```

```python
#Write a function second_largest(numbers) that takes a list of 
# numbers as input and returns the second largest number in the list. If the list has less than two unique numbers, return None.
def second_largest(numbers):
    unique_numbers =list(set(numbers))
    unique_numbers.sort(reverse =True)
    if len(unique_numbers) <2:
        return None
    else:
        return unique_numbers[1]
print(second_largest([30, 25, 50, 25, 70, 18, 49]))
```
