```python
import pandas as pd
```

```python
numbers = pd.Series([1, 2, 3, 4, 5])

numbers
```

```python
print(numbers.dtype)
print(numbers.index)
numbers.values
```

```python
#indexing
print(numbers[3])
numbers[0:4]
```

```python
#location based indexing(iloc)
print(numbers.loc[3])
print(numbers.iloc[0:4])
print(numbers.iloc[[1,2,4]])
```

```python
fruits = ["Apple","Banana", "Mango", "Orange", "Grapes"]
numbers.fruits = fruits
numbers
```

```python
#loc -----> Label based indexing
print(numbers.loc[3])
# In label based indexing both the start and end values are included
```

```python
#Dictionary
fruits_Protiens = {
    "Apple" : 2.0,
    "Banana" : 1.5,
    "Orange" : 1.0,
    "Grape" : 2.5,
    "Guava" : 3.0,

}
s = pd.Series(fruits_Protiens)
s
```

```python
#Conditional Selection
print(s>1)
#print(s[s>1])
```

```python
# Logical Operators ----> and, or, not
print([(s<2) & (s>0.5)])
print([~(s<2)])
```

```python
# Data Frames
data = {
    "Name": ["Bilal", "Ali", "Ahmed", "Hassan", "Hussain"],
    "Age" : [20, 21, 22, "NaN", 24],
    "Department" : ["CS", "IT", "IT", "CS", "IT"],
    "CGPA" : [3.5, 3.6, 3.7, 3.8, 3.9], 
    "Salary" : [50000, 60000, 70000, 80000, 90000]
}
data
```

```python
df = pd.DataFrame(data)
df
```

```python
df.head(2) #Show the first 2 rowa
```

```python
df.tail(3) #Show the last 3 rowa
```

```python
#loc and iloc
df.iloc[0:5]
```

```python
df.loc[0:1, ["Name", "Age"]] 
#In label based indexing both the start and end values are included
```

```python
# Inplace tell us to perfrom operation on original data frame 
# or not
# by default inplace is set to False
#if we set inplace to True then it will perform operation 
# on original data frame and remove the column from original 
# data frame
```

```python
df["Age"] = df["Age"].replace("NaN", 23)
df
```
