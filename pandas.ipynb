```python
import pandas as pd
```

```python
numbers = pd.Series([1, 2, 3, 4, 5])
numbers.name = 'My Numbers'
numbers
```

```python
print(numbers.dtype)
print(numbers.index)
numbers.values
```

```python
#indexing
print(numbers[3])
numbers[0:4]
```

```python
#location based indexing(iloc)
print(numbers.loc[3])
print(numbers.iloc[0:4])
print(numbers.iloc[[1,2,4]])
```

```python
fruits = ["Apple","Banana", "Mango", "Orange", "Grapes"]
numbers.fruits = fruits
numbers
```
