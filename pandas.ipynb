```python
import pandas as pd
```

```python
numbers = pd.Series([1, 2, 3, 4, 5])

numbers
```

```python
print(numbers.dtype)
print(numbers.index)
numbers.values
```

```python
#indexing
print(numbers[3])
numbers[0:4]
```

```python
#location based indexing(iloc)
print(numbers.loc[3])
print(numbers.iloc[0:4])
print(numbers.iloc[[1,2,4]])
```

```python
fruits = ["Apple","Banana", "Mango", "Orange", "Grapes"]
numbers.fruits = fruits
numbers
```

```python
#loc -----> Label based indexing
print(numbers.loc[3])
# In label based indexing both the start and end values are included
```

```python
#Dictionary
fruits_Protiens = {
    "Apple" : 2.0,
    "Banana" : 1.5,
    "Orange" : 1.0,
    "Grape" : 2.5,
    "Guava" : 3.0,

}
s = pd.Series(fruits_Protiens)
s
```

```python
#Conditional Selection
print(s>1)
#print(s[s>1])
```

```python
# Logical Operators ----> and, or, not
print([(s<2) & (s>0.5)])
print([~(s<2)])
```
