Creating a 2D-Array which contain rows and coloumns
It is similar to table or matrix

```python
import numpy as np
```

```python
students = np.array([["Bilal", "Ali", "Ahmad"], [90, 92, 88], ["CS", "SE", "AI"]])
print(students)
print(students.ndim) #Dimension
print(students.shape) #Shape
print(students.size) #Size
print(students.dtype) #Data Type
```

```python
#Accessing complete row
print(students[0]) 
print(students[2])

#Accessing complete column
print(students[:, 0])

#Accessing specific element
#Syntax: array[row_index, column_index]
print(students[0][2])
print(students[2][1])
```

```python
students[0][1]= "Ayesha"
print(students)

#Slicing row
print(students[0:2])
print(students[:, 0:2]) #Slicing column
```

Reshaping An Array

```python
number = np.array([10, 20, 30, 40, 50, 60])
print(number)
matrix = number.reshape(2,3)
print(matrix)
```

Flattening An Array

```python
matrix_1 = np.array([[1,2,3], [4,5,6]])
print(matrix_1)
flat = matrix.flatten()
print(flat)
```
